```python
import ee
import geemap
import pandas as pd
ee.Authenticate
ee.Initialize
```

```python
Map = geemap.Map()

# Panama boundary
countries = ee.FeatureCollection("FAO/GAUL/2015/level0")
panama_fc = countries.filter(ee.Filter.eq("ADM0_NAME", "Panama"))
panama_geom = panama_fc.geometry()

Map.centerObject(panama_geom, 7)
```

### Panama district bounds

```python
districts = ee.FeatureCollection("projects/deforestation-panama/assets/Districts")
```

```python
districts.getInfo()
```

```python
Map.addLayer(districts, {}, 'Districts')
Map
```

### Load PA dataset

```python
features = ee.FeatureCollection("projects/deforestation-495419/assets/panama_protected_areas_polygons").filterBounds(panama_geom)
pa = ee.FeatureCollection("WCMC/WDPA/current/polygons").filterBounds(panama_geom)
```

```python
Map.addLayer(panama_geom, {}, 'Panama Boundary')
# Map.addLayer(pa, {}, 'Protected Areas')
Map.addLayer(features, {}, 'Panama Protected Areas Polygons')

Map
```
